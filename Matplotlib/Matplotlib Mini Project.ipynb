```python
## Use Case: Diabetes Prediction
## Perform Exploratory Data Analysis for the Diabetes Dataset.
## Dataset: Diabetes.csv
## The dataset can be downloaded from https://www.kaggle.com/datasets
## Perform the following tasks
## 1. Load the data in the DataFrame
## 2. Data Pre-processing
## 3. Handle the Categorical Data
## 4. Perform Uni-variate Analysis
## 5. Perform Bi-variate Analysis
```

```python
import pandas as pd
df = pd.read_csv("diabetes.csv")
print(df.shape)
print(df.head())
print(df.info())
```

```python
print(df.isnull().sum())
```

```python
cols_with_zero = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']
df[cols_with_zero] = df[cols_with_zero].replace(0, pd.NA)
print(df.isna().sum())
```

```python
for col in cols_with_zero:
    df[col].fillna(df[col].median(), inplace=True)
```

```python
print(df['Outcome'].value_counts())
```

```python
df = pd.get_dummies(df, drop_first=True)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

df.hist(bins=30, figsize=(15, 10), color='skyblue')
plt.suptitle("Histograms of All Features")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(15, 8))
for i, col in enumerate(df.columns[:-1], 1):  # skip Outcome
    plt.subplot(2, 4, i)
    sns.boxplot(y=df[col], color='lightgreen')
    plt.title(col)
plt.tight_layout()
plt.show()
```

```python
for col in df.columns[:-1]:
    plt.figure(figsize=(6, 4))
    sns.boxplot(x='Outcome', y=col, data=df, palette='Set2')
    plt.title(f'{col} vs Outcome')
    plt.show()
```

```python
plt.figure(figsize=(10, 8))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```

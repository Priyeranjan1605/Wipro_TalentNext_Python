```python
## Perform Exploratory Data Analysis for the dataset Mall_Customers. The dataset can be downloaded from https://www.kaggle.com/datasets
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set(style="whitegrid")
```

```python
df = pd.read_csv("Mall_Customers.csv")
```

```python
print(df.head())
print(df.info())
print(df.describe())
print(df.isnull().sum()) 
```

```python
df.rename(columns={
    'CustomerID': 'Customer_ID',
    'Genre': 'Gender',
    'Age': 'Age',
    'Annual Income (k$)': 'Annual_Income',
    'Spending Score (1-100)': 'Spending_Score'
}, inplace=True)
print(df)
```

```python
# Now performing EDA

plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='Gender', palette='pastel')
plt.title('Gender Distribution')
plt.show()
```

```python
plt.figure(figsize=(8, 4))
sns.histplot(df['Age'], bins=20, kde=True, color='skyblue')
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(8, 4))
sns.histplot(df['Annual_Income'], bins=20, kde=True, color='salmon')
plt.title('Annual Income Distribution')
plt.xlabel('Annual Income (k$)')
plt.show()
```

```python
plt.figure(figsize=(8, 4))
sns.histplot(df['Spending_Score'], bins=20, kde=True, color='lightgreen')
plt.title('Spending Score Distribution')
plt.xlabel('Spending Score (1-100)')
plt.show()
```

```python
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
sns.boxplot(data=df, x='Gender', y='Age', palette='Set2')

plt.subplot(1, 2, 2)
sns.boxplot(data=df, x='Gender', y='Annual_Income', palette='Set2')
plt.tight_layout()
plt.show()
```

```python
sns.pairplot(df.drop('Customer_ID', axis=1), hue='Gender', palette='husl')
plt.suptitle("Pairplot by Gender", y=1.02)
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm', fmt='.2f')
plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x='Age', y='Spending_Score', hue='Gender', palette='Set1')
plt.title('Age vs Spending Score')
plt.show()
```

```python
## Perform Exploratory Data Analysis for the dataset salary_data. The dataset can be downloaded from https://www.kaggle.com/datasets
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set(style='whitegrid')
```

```python
df = pd.read_csv('Salary_Data.csv')
```

```python
print(df.head())
print(df.info())
print(df.describe())
print(df.isnull().sum())  
```

```python
# Now performing EDA

plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Years of Experience', y='Salary', color='blue')
plt.title('Years of Experience vs Salary')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.histplot(df['Years of Experience'], bins=10, kde=True, color='skyblue')
plt.title('Distribution of Years of Experience')
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.histplot(df['Salary'], bins=10, kde=True, color='salmon')
plt.title('Distribution of Salary')
plt.show()
```

```python
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
sns.boxplot(y=df['Years of Experience'], color='lightgreen')
plt.title('Boxplot of Years of Experience')
plt.subplot(1, 2, 2)
sns.boxplot(y=df['Salary'], color='orange')
plt.title('Boxplot of Salary')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(5, 3))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm', fmt='.2f')
plt.title("Correlation Matrix")
plt.show()
```

```python
## Perform Exploratory Data Analysis for the dataset Social Network Ads. The dataset can be downloaded from https://www.kaggle.com/datasets
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
sns.set(style='whitegrid')
```

```python
df = pd.read_csv("Social_Network_Ads.csv")
```

```python
print(df.head())
print(df.info())
print(df.describe())
print(df.isnull().sum())
```

```python
sns.histplot(data=df, x='Age', bins=20, kde=True, color='skyblue')
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()
```

```python
sns.histplot(data=df, x='EstimatedSalary', bins=20, kde=True, color='salmon')
plt.title('Estimated Salary Distribution')
plt.xlabel('Estimated Salary')
plt.ylabel('Count')
plt.show()
```

```python
sns.countplot(x='Purchased', data=df, palette='Set2')
plt.title('Purchase Decision Distribution')
plt.xticks([0, 1], ['Not Purchased', 'Purchased'])
plt.ylabel('Number of Customers')
plt.show()
```

```python
sns.scatterplot(data=df, x='Age', y='EstimatedSalary', hue='Purchased', palette='coolwarm')
plt.title('Age vs Estimated Salary by Purchase Decision')
plt.xlabel('Age')
plt.ylabel('Estimated Salary')
plt.legend(title='Purchased', labels=['No', 'Yes'])
plt.show()
```

```python
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
sns.boxplot(y='Age', data=df, color='lightblue')
plt.title('Boxplot of Age')
plt.subplot(1, 2, 2)
sns.boxplot(y='EstimatedSalary', data=df, color='lightgreen')
plt.title('Boxplot of Estimated Salary')
plt.tight_layout()
plt.show()
```

```python
sns.heatmap(df.corr(), annot=True, cmap='YlGnBu', fmt='.2f')
plt.title("Correlation Matrix")
plt.show()
```
